## Task 9: Fetch records satisfying the following conditions:

### 9a. Tenure > 50 and Gender = 'Female'

```python
condition_a = churn[(churn['tenure'] > 50) & (churn['gender'] == 'Female')]
condition_a
```

### 9b. Gender = 'Male' and SeniorCitizen = 0

```python
condition_b = churn[(churn['gender'] == 'Male') & (churn['SeniorCitizen'] == 0)]
condition_b
```

### 9c. TechSupport = 'Yes' and Churn = 'No'

```python
condition_c = churn[(churn['TechSupport'] == 'Yes') & (churn['Churn'] == 'No')]
condition_c
```

### 9d. Contract type = 'Month-to-month' and Churn = 'Yes'

```python
condition_d = churn[(churn['Contract'] == 'Month-to-month') & (churn['Churn'] == 'Yes')]
condition_d
```

## Task 10: Use a for loop to count customers getting TechSupport and are Male Senior Citizens

```python
count = 0

for index, row in churn.iterrows():
    if row['TechSupport'] == 'Yes' and row['gender'] == 'Male' and row['SeniorCitizen'] == 1:
        count += 1

print(f"Number of customers with TechSupport who are Male Senior Citizens: {count}")
```
